# Pętle w Pythonie

## 1. Po co pętle?

Powtarzanie tych samych instrukcji przez kopiowanie jest niewygodne. Pętla pozwala powtarzać blok bez kopiowania kodu.

```python
print("Próba 1")
print("Próba 2")
print("Próba 3")
print("Próba 4")
print("Próba 5")
print("Próba 6")
print("Próba 7")
print("Próba 8")
print("Próba 9")
print("Próba 10")
```

```python
for i in range(1,11):
    print(f"Próba {i}")
```

**Iteracja** to jedno wykonanie bloku pętli. 

## 2. `while` - dopóki warunek jest prawdziwy

Schemat: `while warunek:` i **wcięty blok**. Warunek sprawdzany jest przed każdą iteracją. Musi istnieć sposób jego zmiany; inaczej program może nigdy się nie zatrzymać.

```python
licznik = 0
while licznik <= 10:
    print("Próba", licznik)
    licznik += 1


print("Po pętli:", licznik)
```

### Śledzenie `while`

| Sprawdzenie | `licznik` | Warunek `<= 3` | Działanie |
|---|---:|---|---|
| 1 | 1 | True | wypisz 1, zwiększ do 2 |
| 2 | 2 | True | wypisz 2, zwiększ do 3 |
| 3 | 3 | True | wypisz 3, zwiększ do 4 |
| 4 | 4 | False | koniec |

**Pytania:** Co się stanie, gdy licznik zacznie od 10? Co jeśli zapomnimy `licznik += 1`?

## 4. `while True` - pętla bezwarunkowa

`while True` ma warunek zawsze prawdziwy. **Kończymy ją świadomie używając `break`**.

```python
while True:
    odpowiedz = input("Wpisz start, aby kontynuować: ")
    if odpowiedz == "start":
        break

    print("Spróbuj ponownie")
    print("coś")
    
    if odpowiedz == "inna":
        print("inna")





print("Rozpoczynamy")
```

## 5. `break` i `continue`

`break` **kończy całą najbliższą pętlę**. `continue` **pomija resztę bieżącej iteracji** i przechodzi do kolejnej.
W `while` trzeba uważać, aby `continue` nie pominęło aktualizacji licznika.

```python
i = 0

while i < 10:
    i += 1

    if i == 3:
        print("continue - wypisano")
        continue

    if i == 5:
        print("Break")
        break

    print(i)

print("Koniec pętli")
```

## 6. `for` — kolejne elementy

`for` pobiera kolejne wartości z sekwencji.
`range(5)` daje 0,1,2,3,4 — **bez 5**.
`range(start, stop, krok)` nie obejmuje wartości `stop`.

```python
for numer in range(1, 6):
    print("Próba", numer)
```

```python
i = 1
while i < 6:
    print(f"Próba {i}")
    i += 1
```

### `range()`

| Wyrażenie | Wynik |
|---|---|
| `range(4)` | 0, 1, 2, 3 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

**Przed uruchomieniem:** przewidź wartości w kolejnej komórce.

```python
for x in range(10, 1, -3):
    print(x)
```

```python
lista = ["Ala", "ma", "kota", "komputer"]
for i in lista:
    print(i)
```

## 7. Jak zbudować `for` za pomocą `while`?

Dla pętli liczbowej `for` można odtworzyć za pomocą `while`:(1) inicjalizacja licznika, (2) warunek końca, (3) wykonanie bloku, (4) zmiana licznika.

```python
print("FOR")
for i in range(2, 9, 2):
    print(i)

print("WHILE")
i = 2
while i < 9:
    print(i)
    i += 2
```

## 8. Liczniki i akumulatory

**Licznik** zwiększamy, gdy wystąpi określone zdarzenie. **Akumulator** gromadzi sumę. Inicjalizujemy je **przed** pętlą. Warunek `if` może znajdować się wewnątrz pętli.

```python
suma = 0
ile_parzystych = 0
for liczba in range(1, 11):
    suma += liczba
    if liczba % 2 == 0:
        ile_parzystych += 1
print("Suma:", suma)
print("Parzyste:", ile_parzystych)
```

## 11. Warunki i pętle w praktyce

Łączymy `while True`, `if`, licznik i akumulator. Poniższy przykład zbiera liczby aż do hasła `koniec`. Najpierw sprawdzamy tekst, **dopiero potem** konwertujemy go na liczbę.

```python
suma = 0
ile = 0
while True:
    wpis = input("Podaj liczbę lub koniec: ")
    if wpis == "koniec":
        break
    liczba = float(wpis)
    suma += liczba
    ile += 1
if ile > 0:
    print("Średnia:", suma / ile)
else:
    print("Nie podano liczb")
```

- `while`: gdy kontynuacja zależy od warunku.
- `while True` + `break`: gdy warunek zakończenia sprawdzamy w środku pętli.
