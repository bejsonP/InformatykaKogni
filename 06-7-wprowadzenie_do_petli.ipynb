# Pętle w Pythonie

## 1. Po co pętle?

Powtarzanie tych samych instrukcji przez kopiowanie jest niewygodne. Pętla pozwala powtarzać blok bez kopiowania kodu.

````python
print("Próba 1")
print("Próba 2")
print("Próba 3")
print("Próba 4")
print("Próba 5")
print("Próba 6")
print("Próba 7")
print("Próba 8")
print("Próba 9")
print("Próba 10")
````

````python
for i in range(1,11):
    print(f"Próba {i}")
````

**Iteracja** to jedno wykonanie bloku pętli. 

## 2. `while` - dopóki warunek jest prawdziwy

Schemat: `while warunek:` i **wcięty blok**. Warunek sprawdzany jest przed każdą iteracją. Musi istnieć sposób jego zmiany; inaczej program może nigdy się nie zatrzymać.

````python
licznik = 0
while licznik <= 10:
    print("Próba", licznik)
    licznik += 1


print("Po pętli:", licznik)
````

### Śledzenie `while`

| Sprawdzenie | `licznik` | Warunek `<= 3` | Działanie |
|---|---:|---|---|
| 1 | 1 | True | wypisz 1, zwiększ do 2 |
| 2 | 2 | True | wypisz 2, zwiększ do 3 |
| 3 | 3 | True | wypisz 3, zwiększ do 4 |
| 4 | 4 | False | koniec |

**Pytania:** Co się stanie, gdy licznik zacznie od 10? Co jeśli zapomnimy `licznik += 1`?

## 4. `while True` - pętla bezwarunkowa

`while True` ma warunek zawsze prawdziwy. **Kończymy ją świadomie używając `break`**.

````python
while True:
    odpowiedz = input("Wpisz start, aby kontynuować: ")
    if odpowiedz == "start":
        break

    print("Spróbuj ponownie")
    print("Coś innego w ramach pętli")
    
    if odpowiedz == "inna":
        print("Wpisano 'inna'")





print("Pętla się skończyła, wykonujemy dalsze elementy kodu")
````

## 5. `break` i `continue`

`break` **kończy całą najbliższą pętlę**. `continue` **pomija resztę bieżącej iteracji** i przechodzi do kolejnej.
W `while` trzeba uważać, aby `continue` nie pominęło aktualizacji licznika.

````python
i = 0

while i < 10:
    i += 1

    if i == 3:
        print("continue - wypisano")
        continue

    if i == 5:
        print("Break")
        break

    print(i)

print("Koniec pętli")
````

## 6. `for` — kolejne elementy

`for` pobiera kolejne wartości z sekwencji.
`range(5)` daje 0,1,2,3,4 — **bez 5**.
`range(start, stop, krok)` nie obejmuje wartości `stop`.

````python
for numer in range(1, 6):
    print("Próba", numer)
````

````python
i = 1
while i < 6:
    print(f"Próba {i}")
    i += 1
````

## 7. Jak zbudować `for` za pomocą `while`?

Dla pętli liczbowej `for` można odtworzyć za pomocą `while`:(1) inicjalizacja licznika, (2) warunek końca, (3) wykonanie bloku, (4) zmiana licznika.

````python
print("FOR")
for i in range(2, 9, 2):
    print(i)

print("WHILE")
i = 2
while i < 9:
    print(i)
    i += 2
````

## 8. Liczniki i akumulatory

**Licznik** zwiększamy, gdy wystąpi określone zdarzenie. **Akumulator** gromadzi sumę. Inicjalizujemy je **przed** pętlą. Warunek `if` może znajdować się wewnątrz pętli.

````python
suma = 0
ile_parzystych = 0
for liczba in range(1, 11):
    suma += liczba
    if liczba % 2 == 0:
        ile_parzystych += 1
print("Suma:", suma)
print("Parzyste:", ile_parzystych)
````

## 9. `for` po napisie

`for` może iterować również po znakach napisu. Nie potrzebujemy `range()`, jeśli interesują nas same znaki.

````python
tekst = "Python"
for znak in tekst:
    print(znak)
````

## 10. Pętle zagnieżdżone

Pętla wewnętrzna wykonuje się od początku dla **każdej iteracji pętli zewnętrznej**. `break` kończy tylko najbliższą pętlę.

````python
for uczestnik in range(1, 4):
    print("Uczestnik", uczestnik)
    for proba in range(1, 3):
        print("  Próba", proba)
````

## 11. Warunki i pętle w praktyce

Łączymy `while True`, `if`, licznik i akumulator. Poniższy przykład zbiera liczby aż do hasła `koniec`. Najpierw sprawdzamy tekst, **dopiero potem** konwertujemy go na liczbę.

````python
suma = 0
ile = 0
while True:
    wpis = input("Podaj liczbę lub koniec: ")
    if wpis == "koniec":
        break
    liczba = float(wpis)
    suma += liczba
    ile += 1
if ile > 0:
    print("Średnia:", suma / ile)
else:
    print("Nie podano liczb")
````

## Podsumowanie

- `while`: gdy kontynuacja zależy od warunku.
- `while True` + `break`: gdy warunek zakończenia sprawdzamy w środku pętli.
- `for` + `range()`: gdy znamy zakres iteracji.
- `continue`: pominięcie reszty jednej iteracji; `break`: zakończenie najbliższej pętli.
- Liczniki i akumulatory inicjalizujemy przed pętlą.
- Zagnieżdżenie: pętla wewnętrzna działa dla każdej iteracji zewnętrznej.

### Ćwiczenie 6. Zgadnij liczbę

Ustal `sekret = 37`. Użytkownik ma maksymalnie pięć prób. Po każdej nietrafionej próbie wypisz `Za mało` albo `Za dużo`. Jeśli trafi, wyświetl numer udanej próby i nie pytaj ponownie. Jeśli nie trafi, pokaż liczbę po wykorzystaniu wszystkich prób.

*Wskazówka:* Sprawdź zachowanie, gdy trafienie następuje w pierwszej i piątej próbie.

````python
sekret = 37

for proba in range(1, 6):
    liczba = int(input("Zgadnij liczbę: "))

    if liczba == sekret:
        print(f"Brawo! Odgadnięto {sekret} w próbie nr {proba}")
        break
    elif liczba < sekret:
        print(f"{liczba} - za mało")
    else:
        print(f"{liczba} - za dużo")
else:
    print("Koniec prób! Szukana liczba to:", sekret)
````

### Ćwiczenie 7. Pomijanie danych

Pobierz **dokładnie sześć** liczb całkowitych. Dodaj tylko liczby dodatnie, a zero i liczby ujemne pomiń za pomocą `continue`. Na końcu wyświetl sumę i liczbę dodatnich wartości. Każde pobranie liczy się jako jedna z sześciu prób.

### Ćwiczenie 8. Średni czas reakcji

Pobieraj czasy reakcji w milisekundach. Wpisanie `0` kończy wprowadzanie danych. Ujemne wartości i czasy przekraczające 1500 ms pomijaj (`continue`). Na końcu podaj liczbę poprawnych pomiarów i ich średnią. Obsłuż brak poprawnych pomiarów.

*Wskazówka:* Zakładamy, że użytkownik podaje liczby całkowite

````python
suma = 0
licznik = 0

while True:
    czas = int(input("Podaj czas reakcji (ms): "))
    print(f"Wprowadzono czas: {czas} ms")

    if czas == 0:
        break

    if czas < 0 or czas > 1500:
        print("Pomijam ostatnią próbę")
        continue

    suma += czas
    licznik += 1

print("Liczba poprawnych pomiarów:", licznik)

if licznik > 0:
    srednia = suma / licznik
    print("Średni czas reakcji:", srednia, "ms")
else:
    print("Brak poprawnych pomiarów")
````

### Ćwiczenie 11. Tabliczka mnożenia

Za pomocą dwóch zagnieżdżonych pętli wyświetl tabliczkę mnożenia od 1 do 5 w pięciu wierszach. Każdy wiersz powinien zawierać pięć wyników oddzielonych spacjami.

*Wskazówka:* Możesz użyć `print(wynik, end=" ")` i osobnego `print()` po zakończeniu wewnętrznej pętli.

#### Parametr `end` w funkcji `print()`

Funkcja `print()` domyślnie kończy wypisywanie znakiem nowej linii (`\n`).

Parametr `end` pozwala określić, **co zostanie wypisane na końcu**, zamiast domyślnego przejścia do nowej linii.

```python
print("A", end=" ")    # Spacja
print("B", end="")     # Brak dodatkowego znaku
print("C", end=" -> ") # Własny separator
print("D")
```

Wynik:
```text
A BC -> D
```

**Najczęściej stosowane wartości:**
- `end="\n"` – nowa linia (domyślne zachowanie)
- `end=" "` – spacja
- `end=""` – brak dodatkowego znaku
- `end=", "` – przecinek i spacja
- `end="\t"` – tabulator

**Przykład zastosowania w pętli:**

```python
for i in range(1, 6):
    print(i, end=" ")
```

Wynik: `1 2 3 4 5`

**Ważne:** `end` jest parametrem funkcji `print()`, a nie osobną instrukcją Pythona. Inne funkcje mogą mieć parametr o tej samej nazwie, ale jego znaczenie zależy od konkretnej funkcji.


````python
for i in range(1, 6):
    for k in range(1, 6):
        print(i * k, end="\t")
    print()
````

### Ćwiczenie 12. Wyniki eksperymentu

W badaniu uczestniczy `3` uczestników. Każdy wykonuje `4` próby. Dla każdej próby pobierz czas reakcji i oblicz średni czas osobno dla każdego uczestnika. Na końcu wyświetl również średnią ze wszystkich 12 prób. Zakładamy, że czasy są dodatnie.

*Wskazówka:* Zastosuj akumulator dla uczestnika oraz osobny akumulator i licznik dla całego badania.

````python
suma_calkowita = 0
licznik_calkowity = 0

for uczestnik in range(1, 4):
    suma_uczestnika = 0

    print("Uczestnik", uczestnik)

    for proba in range(1, 5):
        czas = float(input("Podaj czas reakcji: "))
        print(f" Uczestnik nr {uczestnik}; próba nr {proba}: {czas} ms")

        suma_uczestnika += czas
        suma_calkowita += czas
        licznik_calkowity += 1

    srednia_uczestnika = suma_uczestnika / 4
    print("-- Średni czas uczestnika", uczestnik, ":", srednia_uczestnika, "ms")

srednia_calkowita = suma_calkowita / licznik_calkowity
print(f"Średni czas wszystkich prób: {srednia_calkowita} ms")
````

### Ćwiczenie 13. Liczba cyfr

Pobierz nieujemną liczbę całkowitą i policz liczbę jej cyfr za pomocą `while` i operatora `//`. Nie używaj `str()` ani `len()`. Dla zera wynik powinien wynosić 1.

*Wskazówka:* Każde dzielenie całkowite przez 10 usuwa ostatnią cyfrę.

````python
n = int(input("Podaj nieujemną liczbę całkowitą: "))

licznik = 1

while n >= 10:
    n = n // 10
    licznik += 1

print("Liczba cyfr:", licznik)
````
