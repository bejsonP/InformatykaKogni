# Struktury danych w Pythonie — listy, krotki, zbiory i słowniki

## Po co struktury danych?
Zamiast tworzyć osobną zmienną dla każdego pomiaru, możemy umieścić je w **liście**, czyli obiekcie przechowującym wiele elementów.

```python
t1 = 245
t2 = 280
t3 = 245
czasy = [245, 280, 245]
zmienne = [t1, t2, t3]
print("Każda zmienna osobno: ", t1, t2, t3)
print("Lista liczb: ", czasy)
print("Lista zmiennych t: ", zmienne)

print("\nTyp zmiennej:",type(t1))
print("Typ listy:",type(czasy))
print("Typ listy zmiennych:",type(zmienne))
```

## 1. Listy (`list`)
Lista zachowuje kolejność, pozwala na powtórzenia i można ją zmieniać. Elementy oddzielamy przecinkami, a całość zapisujemy w nawiasach kwadratowych. Pusta lista to `[]`. Lista może zawierać różne typy danych, ale zwykle będziemy chcieli przechowywać jeden typ danych.

```python
czasy = [245, 280, 231, 289, 254]
print(len(czasy))       # len() - pozwala sprawdzić z ilu elementów składa sie obiekt
print(280 in czasy)     # in - pozwala sprawdzić czy element znajduje się w jakimś obiekcie; zwraca wartość logiczną
print(100 not in czasy)
```

```python
# Co ważne len() oraz in można użyć też np. na zmiennych typu str()
print(len("Python"))
print("P" in "Python")
print("p" in "Python")
```

### 1.1. Indeksowanie
Indeks to numer pozycji elementu. Python liczy indeksy od zera. Ujemne indeksy pozwalają liczyć od końca.

| Indeks dodatni | 0 | 1 | 2 | 3 | 4 |
|---|---:|---:|---:|---:|---:|
| Wartość | 245 | 280 | 231 | 289 | 254 |
| Indeks ujemny | -5 | -4 | -3 | -2 | -1 |

Ostatni dodatni indeks to `len(czasy) - 1`. Próba odczytu `czasy[5]` spowoduje `IndexError`.
Dzięki indeksom możemy na bierząco "wyciągać" lub modyfikować poszczególne elementy listy. Robimy to wpisując indeks w nawiasach kwadratowych po naszym obiekcie.

```python
czasy = [245, 280, 231, 289, 254]
print(czasy, end="\n\n")
print("Na indeksie 0:", czasy[0])
print(czasy)
print("Na indeksie -1:", czasy[-1])

print("\nLista dotychczas nie uległa zmianie:\n", czasy)

czasy[1] = 999

print("\nEdycja pierwszego indeksu listy czasy[1] = 999 wygląda tak:\n",czasy)
```

### 1.2. Modyfikowanie listy
Funkcja vs metoda - nazewnictwo to głownie kwestia nomenklatury i przeznaczenia.
Funkcję wywołujemy np. jako `len(czasy)`
Metoda jest operacją dostępną dla danego obiektu, którą wywołujemy po kropce na końcu obiektu np. tak: `czasy.append(300)`.

| Metoda | Działanie |
|---|---|
| `append(x)` | dodaje jeden element na końcu |
| `extend(dane)` | dodaje kolejne elementy kolekcji |
| `remove(x)` | usuwa pierwsze wystąpienie wartości x |
| `pop(n)` | usuwa i zwraca element o indeksie n (bez argumentu: ostatni) |
| `count(x)` | liczy wystąpienia wartości x |

`remove()` zgłosi błąd, jeśli wartości nie ma na liście. Możemy najpierw sprawdzić jej obecność przez `in`.

```python
czasy = [245, 280, 245]
print("Bazowa lista")
print(czasy, end="\n\n\n")

czasy.append(300)
print(czasy, "\t\t<- .append(300)\n")

czasy.extend([210, 260])
print(czasy, "\t<- .extend([210, 260]\n")

print(czasy.count(245), "\t\t\t\t<- .count(245)\n")

czasy.remove(245)
print(czasy, "\t<- .remove(245) usunęło PIERWSZE wystąpienie 245\n")

usuniety = czasy.pop(0)
print(usuniety, "\t\t\t\t<- wartość, którą usunęliśmy przy użyciu czasy.pop(0) (pierwszy element z listy)\n")
print(czasy, "\t\t<- lista po użyciu .pop(0)\n")

czasy.pop()
czasy.pop()
czasy.pop()
print(czasy, "\t\t\t\t<- lista potrzykrotnym użyciu .pop()")
```

### 1.3. Wycinanie (`slicing`)
`lista[start:stop]` wybiera elementy od indeksu start do stop, bez elementu stop. Możemy pominąć granicę, aby sięgnąć do początku lub końca. Trzeci argument oznacza krok, podobnie jak w `range()`.

```python
czasy = [245, 280, 231, 289, 254, 277, 267, 288]
print(czasy[1:4])
print(czasy[:2])
print(czasy[-2:])
print(czasy[::2])
print(czasy[0:5:2])
```

### 1.4. Sortowanie i kopiowanie
`sorted()` zwraca nową posortowaną listę, natomiast `sort()` zmienia istniejącą listę i zwraca `None`.

```python
czasy = [280, 240, 260, 120, 670]
print(czasy, "<- Bazowa")
print(sorted(czasy), "<- Posortowana")
print(czasy, "<- (sorted(czasy) nie zmodyfikowało nam listy 'w miejscu'")
czasy.sort()
print(czasy, "<- .sort() zmodyfikowało liste")
```


Zwykłe przypisanie `b = a` nie kopiuje listy: obie nazwy odnoszą się do tego samego obiektu. Dla prostej listy liczb użyj `copy()`, jeśli chcesz niezależną listę.

```python
a = [10, 20]
print(a)
b = a
b.append(30)
print(a)
c = a.copy()
c.append(40)
print(a)
print(c)
```

## 2. Listy i pętle
Dotychczas `for` pobierała kolejne liczby z `range()`. Jak było sygnalizowane wcześniej - pętle mogą pobierać elementy bezpośrednio z listy. Jeżeli potrzebujemy indeksu, możemy wykorzystać `range(len(lista))`.

```python
czasy = [245, 280, 231]
for czas in czasy:
    print(czas)

for i in range(len(czasy)):
    print(i, czasy[i])
```

### Filtrowanie używając list
Tworzymy pustą listę, a następnie dodajemy do niej tylko wartości spełniające warunek. Za poprawny pomiar przyjmijmy umownie czas od 150 do 1500 ms.

```python
czasy = [245, 0, 280, -1, 1800, 231]
poprawne = []
for czas in czasy:
    if 150 <= czas <= 1500:
        poprawne.append(czas)
print(poprawne)
if len(poprawne) > 0:
    print('Średnia:', sum(poprawne) / len(poprawne))
else:
    print('Brak danych')
```

`sum(lista)` dodaje liczby, `min(lista)` i `max(lista)` wybierają skrajne wartości. Średnią liczymy tylko dla niepustej listy. `min()` i `max()` również wymagają niepustej kolekcji, jeśli nie podamy wartości domyślnej.

## Krotki (`tuple`)
Krotka zachowuje kolejność i duplikaty, lecz nie można podmienić jej elementów po utworzeniu. Przydaje się do zapisu ustalonego zestawu informacji, np. współrzędnych. Najczęściej używamy nawiasów okrągłych. W krotce jednoelementowej konieczny jest przecinek.

```python
bodziec = (120, 80)
print(bodziec[0])
print(len(bodziec))
print(type((5)))
print(type((5,)))
# bodziec[0] = 200  # TypeError: nie można zmienić elementu krotki
```

### Rozpakowywanie krotek
Każdy element można przypisać do osobnej zmiennej. Liczba zmiennych musi odpowiadać liczbie elementów.

```python
bodziec = (120, 80)
x, y = bodziec
print(x, y)

proby = [('U01', 245), ('U02', 280)]
for identyfikator, czas in proby:
    print(identyfikator, czas)
```

## Zbiory (`set`)
Zbiór przechowuje unikalne elementy. Nie ma indeksów ani gwarantowanej kolejności iterowania. Pusty zbiór tworzymy przez `set()` (zapis `{}` tworzy słownik). Zbiór można utworzyć z listy, aby usunąć powtórzenia.

```python
odpowiedzi = ['A', 'B', 'A', 'C']
unikalne = set(odpowiedzi)
print(unikalne)
print(len(unikalne))
print('A' in unikalne)
unikalne.add('D')
unikalne.discard('B')
print(unikalne)
```

### Operacje na zbiorach
| Operator | Znaczenie |
|---|---|
| `A \| B` | suma: elementy z co najmniej jednego zbioru |
| `A & B` | część wspólna |
| `A - B` | elementy A, których nie ma w B |

Wyniki mogą zostać wyświetlone w innej kolejności.

```python
a = {'U01', 'U02', 'U03'}
b = {'U02', 'U04'}
print(a | b)
print(a & b)
print(a - b)
```

## 6. Słowniki (`dict`)
Słownik przechowuje pary **klucz: wartość**. Klucze są unikalne, wartości mogą się powtarzać. Odczytujemy dane przez klucz zamiast indeksu pozycji. W Pythonie 3.7 i nowszym słowniki zachowują kolejność dodawania kluczy.

```python
uczestnik = {'id': 'U01', 'wiek': 22, 'czas_ms': 245}
print(uczestnik['wiek'])
uczestnik['czas_ms'] = 250
uczestnik['grupa'] = 'A'
print(uczestnik)
print('wiek' in uczestnik)
print(uczestnik.get('wynik', 'Brak danych'))
```

Brak klucza przy odczycie `slownik[klucz]` powoduje `KeyError`. Metoda `get(klucz, domyslna)` pozwala podać wartość zastępczą. `in` sprawdza klucze, a nie wartości. `items()` umożliwia iterowanie po parach klucz–wartość.

```python
uczestnik = {'id': 'U01', 'wiek': 22, 'czas_ms': 245}
for klucz, wartosc in uczestnik.items():
    print(klucz, wartosc)
```

### Zliczanie odpowiedzi
Pusty słownik `{}` będzie gromadził liczbę wystąpień każdej odpowiedzi. Jeśli klucza jeszcze nie ma, zaczynamy od 1. W przeciwnym razie zwiększamy istniejącą wartość.

```python
odpowiedzi = ['A', 'B', 'A', 'C', 'B', 'A']
liczebnosci = {}
for odpowiedz in odpowiedzi:
    if odpowiedz in liczebnosci:
        liczebnosci[odpowiedz] += 1
    else:
        liczebnosci[odpowiedz] = 1
print(liczebnosci)
```
