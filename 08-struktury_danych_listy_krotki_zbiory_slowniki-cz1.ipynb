# Struktury danych w Pythonie — listy, krotki, zbiory i słowniki

## Po co struktury danych?
Zamiast tworzyć osobną zmienną dla każdego pomiaru, możemy umieścić je w **liście**, czyli obiekcie przechowującym wiele elementów.

```python
t1 = 245
t2 = 280
t3 = 245
czasy = [245, 280, 245]
zmienne = [t1, t2, t3]
print("Każda zmienna osobno: ", t1, t2, t3)
print("Lista liczb: ", czasy)
print("Lista zmiennych t: ", zmienne)

print("\nTyp zmiennej:",type(t1))
print("Typ listy:",type(czasy))
print("Typ listy zmiennych:",type(zmienne))
```

## 1. Listy (`list`)
Lista zachowuje kolejność, pozwala na powtórzenia i można ją zmieniać. Elementy oddzielamy przecinkami, a całość zapisujemy w nawiasach kwadratowych. Pusta lista to `[]`. Lista może zawierać różne typy danych, ale zwykle będziemy chcieli przechowywać jeden typ danych.

```python
czasy = [245, 280, 231, 289, 254]
print(len(czasy))       # len() - pozwala sprawdzić z ilu elementów składa sie obiekt
print(280 in czasy)     # in - pozwala sprawdzić czy element znajduje się w jakimś obiekcie; zwraca wartość logiczną
print(100 not in czasy)
```

```python
# Co ważne len() oraz in można użyć też np. na zmiennych typu str()
print(len("Python"))
print("P" in "Python")
print("p" in "Python")
```

### 1.1. Indeksowanie
Indeks to numer pozycji elementu. Python liczy indeksy od zera. Ujemne indeksy pozwalają liczyć od końca.

| Indeks dodatni | 0 | 1 | 2 | 3 | 4 |
|---|---:|---:|---:|---:|---:|
| Wartość | 245 | 280 | 231 | 289 | 254 |
| Indeks ujemny | -5 | -4 | -3 | -2 | -1 |

Ostatni dodatni indeks to `len(czasy) - 1`. Próba odczytu `czasy[5]` spowoduje `IndexError`.
Dzięki indeksom możemy na bierząco "wyciągać" lub modyfikować poszczególne elementy listy. Robimy to wpisując indeks w nawiasach kwadratowych po naszym obiekcie.

```python
czasy = [245, 280, 231, 289, 254]
print(czasy, end="\n\n")
print("Na indeksie 0:", czasy[0])
print(czasy)
print("Na indeksie -1:", czasy[-1])

print("\nLista dotychczas nie uległa zmianie:\n", czasy)

czasy[1] = 999

print("\nEdycja pierwszego indeksu listy czasy[1] = 999 wygląda tak:\n",czasy)
```

### 1.2. Modyfikowanie listy
Funkcja vs metoda - nazewnictwo to głownie kwestia nomenklatury i przeznaczenia.
Funkcję wywołujemy np. jako `len(czasy)`
Metoda jest operacją dostępną dla danego obiektu, którą wywołujemy po kropce na końcu obiektu np. tak: `czasy.append(300)`.

| Metoda | Działanie |
|---|---|
| `append(x)` | dodaje jeden element na końcu |
| `extend(dane)` | dodaje kolejne elementy kolekcji |
| `remove(x)` | usuwa pierwsze wystąpienie wartości x |
| `pop(n)` | usuwa i zwraca element o indeksie n (bez argumentu: ostatni) |
| `count(x)` | liczy wystąpienia wartości x |

`remove()` zgłosi błąd, jeśli wartości nie ma na liście. Możemy najpierw sprawdzić jej obecność przez `in`.

```python
czasy = [245, 280, 245]
print("Bazowa lista")
print(czasy, end="\n\n\n")

czasy.append(300)
print(czasy, "\t\t<- .append(300)\n")

czasy.extend([210, 260])
print(czasy, "\t<- .extend([210, 260])\n")

print(czasy.count(245), "\t\t\t\t<- .count(245)\n")

czasy.remove(245)
print(czasy, "\t<- .remove(245) usunęło PIERWSZE wystąpienie 245\n")

usuniety = czasy.pop(0)
print(usuniety, "\t\t\t\t<- wartość, którą usunęliśmy przy użyciu czasy.pop(0) (pierwszy element z listy)\n")
print(czasy, "\t\t<- lista po użyciu .pop(0)\n")

czasy.pop()
czasy.pop()
czasy.pop()
print(czasy, "\t\t\t\t<- lista potrzykrotnym użyciu .pop()")
```

```python
czasy = [245, 280, 245]
czasy.append([210, 260])    # dodajemy 1 element do listy (cała lista [210, 260] traktowana jest jako 1 element)
print(czasy, len(czasy))

czasy = [245, 280, 245]
czasy.extend([210, 260])    # dodajemy zawartość jednej listy do drugiej
print(czasy, len(czasy))
```

#### Co jeżeli chcemy usunąć elementy danego typu?
np. mamy listę zawierającą zarówno liczbym jak i stringi i chcemy usunąć z niej wszystkie zmienne tekstowe

```python
lista = [12, "tekst", 25, "tekst2", 30, "ABC"]

for element in lista.copy():
    if type(element) == str:
        lista.remove(element)

print(lista)
```

### 1.3. Wycinanie (`slicing`)
`lista[start:stop]` wybiera elementy od indeksu start do stop, bez elementu stop. Możemy pominąć granicę, aby sięgnąć do początku lub końca. Trzeci argument oznacza krok, podobnie jak w `range()`.

```python
czasy = [245, 280, 231, 289, 254, 277, 267, 288]
print(czasy[1:4])
print(czasy[:2])
print(czasy[-2:])
print(czasy[::2])
print(czasy[0:5:2])
```

### 1.4. Sortowanie
`sorted()` zwraca nową posortowaną listę, natomiast `sort()` zmienia istniejącą listę i zwraca `None`.

```python
czasy = [280, 240, 260, 120, 670]
print(czasy, "<- Bazowa")
print(sorted(czasy), "<- Posortowana")
print(czasy, "<- (sorted(czasy) nie zmodyfikowało nam listy 'w miejscu'")
czasy.sort()
print(czasy, "<- .sort() zmodyfikowało liste")
```

### 1.5 Kopiowanie
Zwykłe przypisanie `b = a` nie kopiuje listy: obie nazwy odnoszą się do tego samego obiektu. Dla prostej listy liczb użyj `copy()`, jeśli chcesz niezależną listę.

```python
a = [10, 20]
print(a)
b = a
b.append(30)
print(a)
c = a.copy()
c.append(40)
print(a)
print(c)
```

## Ćwiczenie 3. Łączenie list - różnice miedzy append() i extend()
Dane są dwie listy zawierające wyniki pomiarów z dwóch sesji eksperymentalnych.

Utwórz trzy niezależne kopie listy sesja_1, a następnie:
1. Do pierwszej kopii dodaj całą listę sesja_2 za pomocą metody append().
2. Do drugiej kopii dodaj listę sesja_2 za pomocą metody extend().
3. Do trzeciej kopii dodaj elementy listy sesja_2, wykorzystując pętlę oraz metodę append(), tak aby uzyskać taki sam efekt jak w przypadku metody extend().

Wyświetl zawartość oraz długość każdej z trzech list. Porównaj drugą i trzecią listę za pomocą operatora ==. Wyjaśnij, skąd wynikają różnice między otrzymanymi listami.

```python
sesja_1 = [240, 280, 310]
sesja_2 = [250, 290, 270]

# Rozwiązanie:
sesja_1 = [240, 280, 310]
sesja_2 = [250, 290, 270]

# Utworzenie trzech niezależnych kopii listy
kopia_1 = sesja_1.copy()
kopia_2 = sesja_1.copy()
kopia_3 = sesja_1.copy()

# 1. Dodanie całej listy za pomocą append()
kopia_1.append(sesja_2)

print("Lista 1:", kopia_1)
print("Długość:", len(kopia_1))

# 2. Rozszerzenie listy za pomocą extend()
kopia_2.extend(sesja_2)

print("Lista 2:", kopia_2)
print("Długość:", len(kopia_2))

# 3. Odtworzenie działania extend() za pomocą append() i pętli
for element in sesja_2:
    kopia_3.append(element)

print("Lista 3:", kopia_3)
print("Długość:", len(kopia_3))

# Porównanie otrzymanych list
print("Czy lista 2 i lista 3 są identyczne?", kopia_2 == kopia_3)
```

## Ćwiczenie 5. Sortowanie i kopiowanie
Dana jest lista zawierająca wyniki punktowe uczestników quizu.

Utwórz nową listę wyników posortowaną rosnąco, nie modyfikując oryginalnej listy. Następnie wyświetl trzy najniższe oraz trzy najwyższe wyniki.

Utwórz niezależną kopię oryginalnej listy, usuń z niej pierwszy element i posortuj ją za pomocą metody sort(). Wyświetl wszystkie trzy listy.

```python
punkty = [14, 8, 19, 5, 12, 20, 16]

# Rozwiązanie:

posortowane = sorted(punkty) # Utworzenie nowej posortowanej listy

# Wyświetlenie trzech najniższych i najwyższych wyników
print("Najniższe wyniki:", posortowane[:3])
print("Najwyższe wyniki:", posortowane[-3:])

# Utworzenie niezależnej kopii
kopia = punkty.copy()

kopia.pop(0) # Usunięcie pierwszego elementu

kopia.sort() # Sortowanie kopii

# Wyświetlenie wszystkich list
print("Oryginalna lista:", punkty)
print("Posortowana lista:", posortowane)
print("Zmodyfikowana kopia:", kopia)
```

## Ćwiczenie 6. Filtrowanie pomiarów
Lista zawiera czasy reakcji zarejestrowane podczas eksperymentu. Za poprawne uznaj pomiary mieszczące się w przedziale od 150 do 1000 ms włącznie.

Za pomocą pętli usuń wszystkie niepoprawne pomiary. Następnie posortuj pozostałe pomiary rosnąco i wyświetl ich liczbę oraz trzy najkrótsze czasy reakcji.

```python
czasy = [240, 0, 120, 280, 150, 1300, 310, -5, 1000, 210]

# Usunięcie niepoprawnych pomiarów
# Uwaga: wykorzystujemy czasy.copy() w pętli, aby uniknąć pomijania elementów podczas usuwania wartości z oryginalnej listy.
for czas in czasy.copy():
    if czas < 150 or czas > 1000:
        czasy.remove(czas)

# Sortowanie pozostałych pomiarów
czasy.sort()

# Wyświetlenie wyników
print("Poprawne pomiary:", czasy)
print("Liczba poprawnych pomiarów:", len(czasy))
print("Trzy najkrótsze czasy reakcji:", czasy[:3])
```

## Ćwiczenie 7. Wprowadzanie i przetwarzanie danych
Napisz program, który pobierze od użytkownika liczbę całkowitą n. Jeżeli n jest mniejsze lub równe 0, program wypisze „Błąd”. W przeciwnym przypadku za pomocą pętli pobierze n kolejnych liczb całkowitych i zapisze je w liście.

Następnie program powinien:
- wyświetlić listę wprowadzonych liczb,
- utworzyć niezależną kopię listy,
- usunąć z kopii wszystkie liczby ujemne,
- posortować zmodyfikowaną listę rosnąco,
- wyświetlić zmodyfikowaną listę oraz jej długość.
Oryginalna lista nie powinna zostać zmieniona.

```python
n = int(input("Podaj liczbę elementów: "))

if n <= 0:
    print("Błąd")

else:
    liczby = []
    
    # Pobranie n liczb i zapisanie ich w liście
    for i in range(n):
        liczba = int(input("Podaj liczbę: "))
        liczby.append(liczba)

    # Wyświetlenie oryginalnej listy
    print("Wprowadzone liczby:", liczby)

    # Utworzenie niezależnej kopii
    kopia = liczby.copy()

    # Usunięcie liczb ujemnych
    for liczba in liczby:
        if liczba < 0:
            kopia.remove(liczba)

    # Sortowanie zmodyfikowanej listy
    kopia.sort()

    # Wyświetlenie wyników
    print("Zmodyfikowana lista:", kopia)
    print("Długość zmodyfikowanej listy:", len(kopia))
```
