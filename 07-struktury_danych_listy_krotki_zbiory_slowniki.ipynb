## 1. Listy — podstawy

Gdy mamy wiele pomiarów, nie chcemy tworzyć osobnej zmiennej dla każdego pomiaru. Lista `list` przechowuje uporządkowaną kolekcję wartości, może mieć duplikaty i można ją zmieniać.

Indeksy zaczynają się od **0**; indeks `-1` oznacza ostatni element.

```python
czas1 = 245
czas2 = 267
czas3 = 231

czasy = [245, 267, 231, 289, 254]
print(czasy)
print("Pierwszy:", czasy[0])
print("Ostatni:", czasy[-1])
print("Liczba pomiarów:", len(czasy))
```

### Tworzenie list i dostęp do elementów

Listy mogą zawierać różne typy danych, ale zwykle używamy elementów jednego typu. Pusta lista to `[]`.

```python
oceny = [3, 4, 5, 4]
pusta = []

print(oceny[1])
print(len(pusta))
print(4 in oceny)       # czy wartość występuje na liście?
print(oceny.count(4))  # ile razy występuje?
```

### Modyfikowanie list

- `append(x)` — dodaje **jeden element** na końcu.
- `extend(iterowalne)` — dodaje kolejne elementy.
- `insert(i, x)` — wstawia element pod indeks `i`.
- `remove(x)` — usuwa **pierwsze wystąpienie wartości** `x` (zgłasza błąd, gdy brak wartości).
- `pop(i)` — usuwa i **zwraca** element o indeksie `i`; bez argumentu usuwa ostatni.
- `len()`, `count()`, `index()`, `sort()`, `reverse()` — dodatkowe operacje.

Uwaga: `append()`, `remove()` i `pop()` w liście działają **inaczej** niż operacje o tych samych nazwach w innych strukturach.

```python
wyniki = [240, 260, 280]
wyniki.append(300)
wyniki.extend([310, 320])
wyniki.insert(1, 250)
print("Po dodawaniu:", wyniki)

wyniki[0] = 245
wyniki.remove(280)       # usuwa wartość, a nie indeks
usuniety = wyniki.pop(1) # usuwa element o indeksie 1 i go zwraca
print("Usunięto:", usuniety)
print("Po zmianach:", wyniki)
```

```python
liczby = [5, 2, 5, 8, 1]
print("Liczba piątek:", liczby.count(5))
print("Pierwszy indeks 5:", liczby.index(5))
liczby.sort()
print("Rosnąco:", liczby)
liczby.reverse()
print("Odwrócona kolejność:", liczby)
```

## 2. Krotki, zbiory i słowniki — cechy i różnice

| Typ | Zapis | Kolejność | Duplikaty | Modyfikacja | Dostęp |
|---|---|---|---|---|---|
| Lista (`list`) | `[1, 2]` | Zachowana | Tak | Tak | Indeksy |
| Krotka (`tuple`) | `(1, 2)` | Zachowana | Tak | Nie | Indeksy |
| Zbiór (`set`) | `{1, 2}` | Brak gwarantowanej kolejności | Nie | Tak: dodaj/usuń | Przynależność (`in`), bez indeksów |
| Słownik (`dict`) | `{"a": 1}` | Kolejność wstawiania zachowana | Klucze unikalne, wartości mogą się powtarzać | Tak | Klucze |

**Ważne:** „zbiór” (`set`) to konkretna struktura Pythona; „struktury danych” to szersze pojęcie obejmujące także listy, krotki i słowniki.

### 2.1. Krotki (`tuple`)

Krotka przypomina listę, ale nie można podmienić jej elementów, użyć `append()`, `remove()` ani `pop()`. Dostępne są m.in. `len()`, `count()`, `index()` i indeksowanie.

```python
wspolrzedne = (10, 20)
kolory = ("czerwony", "zielony", "czerwony")
print(wspolrzedne[0])
print(kolory.count("czerwony"))
print(kolory.index("zielony"))
print("Długość:", len(wspolrzedne))

# wspolrzedne[0] = 15  # TypeError: tuple nie pozwala na taką zmianę
# wspolrzedne.append(30)  # AttributeError: brak metody append
```

### 2.2. Zbiory (`set`)

Zbiór przechowuje **unikalne elementy**. Nie pozwala na indeksowanie. `add(x)` dodaje element, `update(iterowalne)` dodaje wiele, `remove(x)` usuwa wartość (błąd, gdy nie istnieje), `discard(x)` usuwa bez błędu, a `pop()` usuwa i zwraca **dowolny** element — nie ostatni!

Typowe operacje matematyczne: suma `|`, część wspólna `&`, różnica `-`.

```python
odpowiedzi = {"tak", "nie", "tak"}
print("Zbiór:", odpowiedzi)  # kolejność wyświetlania może być różna
odpowiedzi.add("nie wiem")
odpowiedzi.update(["tak", "brak odpowiedzi"])
print("Po dodaniu:", odpowiedzi)
odpowiedzi.remove("nie")
odpowiedzi.discard("nie")  # brak błędu przy ponownym usunięciu
print("Po usunięciu:", odpowiedzi)

# odpowiedzi.append("nowa")  # AttributeError: set nie ma append
# print(odpowiedzi[0])       # TypeError: nie można indeksować set
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5}
print("Suma zbiorów:", a | b)
print("Część wspólna:", a & b)
print("Różnica a - b:", a - b)
print("3 należy do a:", 3 in a)

# Metoda pop() w set usuwa dowolny element, nie ma pojęcia "ostatniego".
roboczy = {10, 20, 30}
element = roboczy.pop()
print("Usunięty element:", element, "; pozostało:", roboczy)
```

### 2.3. Słowniki (`dict`)

Słownik przechowuje pary **klucz → wartość**. Dostęp do danych odbywa się przez klucz, nie przez numer indeksu. `d[key] = value` dodaje lub aktualizuje wpis, `update()` może aktualizować wiele pozycji, a `pop(key)` usuwa **wskazany klucz** i zwraca jego wartość. `get(key, domyślna)` pozwala bezpiecznie odczytywać brakujące klucze. Nie ma metod `append()` i `remove()`.

```python
uczestnik = {"wiek": 22, "czas_ms": 254, "poprawne": 18}
print("Wiek:", uczestnik["wiek"])
print("Brakujące pole:", uczestnik.get("grupa", "nie podano"))

uczestnik["czas_ms"] = 248
uczestnik["grupa"] = "A"
uczestnik.update({"poprawne": 19})
print("Klucze:", list(uczestnik.keys()))
print("Wartości:", list(uczestnik.values()))
print("Pary:", list(uczestnik.items()))

usunieta_wartosc = uczestnik.pop("grupa")
print("Usunięta wartość:", usunieta_wartosc)
print("Po usunięciu:", uczestnik)

# uczestnik.append(10)     # AttributeError: dict nie ma append
# uczestnik.remove("wiek") # AttributeError: dict nie ma remove
```

### Co oznacza `pop()` w różnych strukturach?

- `lista.pop(indeks)` — usuwa i zwraca element o **indeksie** (domyślnie ostatni).
- `zbior.pop()` — usuwa i zwraca **dowolny** element; nie przyjmuje indeksu.
- `slownik.pop(klucz)` — usuwa wpis o **kluczu** i zwraca skojarzoną wartość.
- `krotka` — brak `pop()`.

Nie należy zakładać, że ta sama nazwa metody oznacza identyczne zachowanie.

## 3. Listy i pętle

Najpierw przejdziemy przez listę za pomocą poznanego wcześniej `while`, potem wprowadzimy `for` jako wygodniejszy sposób iterowania po elementach. `range()` służy do iterowania po liczbach, a `enumerate()` daje jednocześnie indeks i element.

```python
czasy = [245, 267, 231, 289]

i = 0
while i < len(czasy):
    print(f"Pomiar {i + 1}: {czasy[i]} ms")
    i += 1
```

```python
czasy = [245, 267, 231, 289]

for czas in czasy:
    print(f"Czas: {czas} ms")

print("--- z indeksem ---")
for i, czas in enumerate(czasy, start=1):
    print(f"Pomiar {i}: {czas} ms")
```

```python
# Pętle po liczbach: range(start, stop, step)
for i in range(1, 6):
    print(i)

# Równoważna konstrukcja z while:
i = 1
while i < 6:
    print(i)
    i += 1
```

### Akumulator i filtrowanie warunkowe

```python
czasy = [245, 267, 231, 289, 254]
suma = 0

for czas in czasy:
    suma += czas

srednia = suma / len(czasy)
print(f"Średnia: {srednia:.2f} ms")

wolniejsze = []
for czas in czasy:
    if czas > 250:
        wolniejsze.append(czas)

print("Powyżej 250 ms:", wolniejsze)
```

### Pętla po słowniku i zagnieżdżone pętle

```python
uczestnik = {"wiek": 22, "czas_ms": 254, "poprawne": 18}
for klucz, wartosc in uczestnik.items():
    print(f"{klucz}: {wartosc}")

wyniki_blokow = [[240, 250], [270, 260], [230, 245]]
for numer_bloku, blok in enumerate(wyniki_blokow, start=1):
    for czas in blok:
        print(f"Blok {numer_bloku}: {czas} ms")
```

## 4. Wycinanie (`slicing`)

Składnia: `kolekcja[start:stop:step]`. Indeks `stop` jest **wyłączony**. Przy pominiętych granicach Python przyjmuje początek/koniec zgodnie z kierunkiem kroku. Wycinanie działa m.in. na listach, krotkach i napisach, ale **nie na zbiorach ani słownikach**.

```python
czasy = [210, 220, 230, 240, 250, 260]
print("Pierwsze trzy:", czasy[:3])
print("Od indeksu 2:", czasy[2:])
print("Indeksy 1, 2, 3:", czasy[1:4])
print("Co drugi:", czasy[::2])
print("Od końca:", czasy[::-1])

wspolrzedne = (1, 2, 3, 4)
print("Wycięta krotka:", wspolrzedne[1:3])
print("Fragment napisu:", "eksperyment"[:5])
```

### Uwaga: wycinek listy tworzy nową listę

Modyfikacja wycinka nie zmienia oryginalnej listy, natomiast zwykłe przypisanie drugiej nazwy do tej samej listy nie tworzy kopii.

```python
a = [10, 20, 30]
b = a[:]         # płytka kopia listy
c = a            # druga nazwa tej samej listy
b[0] = 99
c[1] = 77
print("a:", a)   # [10, 77, 30]
print("b:", b)   # [99, 20, 30]
```

## 5. Praktyczne przetwarzanie danych

Przykłady łączą listy, słowniki, warunki i pętle. Są rozwiązane, aby pokazać kompletny tok postępowania.

### 5.1. Walidacja i analiza pomiarów

```python
pomiary = [245, -10, 280, 0, 260, 310, -5]
poprawne = []
odrzucone = 0

for czas in pomiary:
    if czas <= 0:
        odrzucone += 1
        continue
    poprawne.append(czas)

print("Poprawne pomiary:", poprawne)
print("Odrzucone:", odrzucone)
if poprawne:
    print(f"Średnia: {sum(poprawne) / len(poprawne):.2f} ms")
    print("Minimum:", min(poprawne), "ms")
    print("Maksimum:", max(poprawne), "ms")
else:
    print("Brak poprawnych pomiarów")
```

### 5.2. Zliczanie odpowiedzi za pomocą słownika

```python
odpowiedzi = ["tak", "nie", "tak", "nie wiem", "tak", "nie"]
liczniki = {}

for odp in odpowiedzi:
    liczniki[odp] = liczniki.get(odp, 0) + 1

for odp, ile in liczniki.items():
    print(f"{odp}: {ile}")
```

### 5.3. Lista słowników — analiza grupy

```python
uczestnicy = [
    {"id": "U01", "wiek": 21, "czas_ms": 245},
    {"id": "U02", "wiek": 24, "czas_ms": 280},
    {"id": "U03", "wiek": 20, "czas_ms": 230},
]

suma = 0
for osoba in uczestnicy:
    suma += osoba["czas_ms"]
    print(f"{osoba['id']}: {osoba['czas_ms']} ms")

print(f"Średnia grupy: {suma / len(uczestnicy):.2f} ms")
```

## 6. Rozwiązane symulacje porównawcze — jaki „pojemnik” wybrać?

Poniższe komórki można uruchamiać i modyfikować. Pokazują realne różnice: duplikaty, kolejność, mutowalność, sposób dostępu oraz znaczenie `pop()` / `remove()` / `append()`. Nie są to zadania do samodzielnego uzupełnienia — zawierają kompletne rozwiązania.

### Symulacja A — te same odpowiedzi w czterech strukturach

**Pytanie:** co dzieje się z powtarzającymi się wartościami i jak je odczytujemy?

```python
dane = ["tak", "nie", "tak", "nie wiem"]
lista = list(dane)
krotka = tuple(dane)
zbior = set(dane)
slownik = {f"proba_{i}": odp for i, odp in enumerate(dane, start=1)}

print("Lista:", lista, "| liczba elementów:", len(lista))
print("Krotka:", krotka, "| liczba elementów:", len(krotka))
print("Zbiór:", zbior, "| liczba elementów:", len(zbior))
print("Słownik:", slownik, "| liczba par:", len(slownik))
print("Druga odpowiedź na liście:", lista[1])
print("Druga odpowiedź w krotce:", krotka[1])
print("Druga próba ze słownika:", slownik["proba_2"])
print("Czy 'tak' należy do zbioru?", "tak" in zbior)
# Nie ma zbior[1]: set nie ma zdefiniowanej kolejności ani indeksów.
```

### Symulacja B — dodaj i usuń: podobne intencje, różne metody

Dla każdej struktury wykonujemy odpowiedniki dodawania i usuwania. `pop()` ma różne argumenty i znaczenia.

```python
lista = ["A", "B", "C"]
lista.append("D")
lista.remove("B")
print("LISTA — pop zwraca ostatni element:", lista.pop(), "| reszta:", lista)

krotka = ("A", "B", "C")
print("KROTKA — bez append/remove/pop:", krotka)
# Aby uzyskać nową krotkę, można połączyć dwie:
krotka = krotka + ("D",)
print("KROTKA — nowa wartość:", krotka)

zbior = {"A", "B", "C"}
zbior.add("D")
zbior.remove("B")
print("ZBIÓR — po add/remove:", zbior)
print("ZBIÓR — pop zwraca dowolny element:", zbior.pop(), "| reszta:", zbior)

slownik = {"A": 1, "B": 2, "C": 3}
slownik["D"] = 4
print("SŁOWNIK — pop zwraca wartość pod kluczem B:", slownik.pop("B"), "| reszta:", slownik)
```

### Symulacja C — historia pomiarów kontra unikalne identyfikatory

Lista zachowuje powtórzenia i kolejność prób. Zbiór przydaje się do sprawdzania unikalności identyfikatorów. Słownik może przypisywać identyfikatorowi wynik, ale powtórzenie klucza nadpisze poprzednią wartość.

```python
identyfikatory = ["U01", "U02", "U01", "U03", "U02"]
print("Wszystkie wpisy:", identyfikatory)
print("Liczba wpisów:", len(identyfikatory))
print("Liczba unikalnych identyfikatorów:", len(set(identyfikatory)))

wyniki = {}
wyniki["U01"] = 245
wyniki["U02"] = 268
wyniki["U01"] = 238  # nadpisuje wcześniejszy wynik U01
print("Ostatnie wyniki per uczestnik:", wyniki)
print("Jeśli chcemy zachować historię, wybieramy np. listę pomiarów lub słownik list.")

historia = {}
for uid, czas in [("U01", 245), ("U02", 268), ("U01", 238)]:
    if uid not in historia:
        historia[uid] = []
    historia[uid].append(czas)
print("Historia:", historia)
```

### Symulacja D — wybór struktury do konkretnego problemu

- **Kolejne pomiary, powtórzenia, indeksy** → lista.
- **Stały zestaw współrzędnych** → krotka.
- **Unikalne elementy, część wspólna** → zbiór.
- **Wyszukiwanie wyniku po identyfikatorze** → słownik.

Jedno zadanie może wymagać **połączenia struktur** (np. lista słowników albo słownik list).

```python
# Lista słowników — zachowujemy kolejność wszystkich prób.
proby = [
    {"osoba": "U01", "czas": 245},
    {"osoba": "U02", "czas": 268},
    {"osoba": "U01", "czas": 238},
]

# Zbiór — szybko otrzymujemy unikalne identyfikatory.
unikalni = set()
for proba in proby:
    unikalni.add(proba["osoba"])

# Słownik list — zachowujemy wszystkie pomiary każdej osoby.
pomiary_osob = {}
for proba in proby:
    osoba = proba["osoba"]
    if osoba not in pomiary_osob:
        pomiary_osob[osoba] = []
    pomiary_osob[osoba].append(proba["czas"])

print("Kolejne próby:", proby)
print("Unikalne osoby:", unikalni)
print("Historia każdej osoby:", pomiary_osob)
for osoba, czasy in pomiary_osob.items():
    print(f"{osoba}: średnia {sum(czasy) / len(czasy):.2f} ms")
```
