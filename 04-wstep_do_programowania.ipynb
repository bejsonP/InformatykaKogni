# Zmienne

**Zmienna** to nazwa, pod którą przechowywana jest wartość.

Operator `=` oznacza **przypisanie wartości**.

```python
x = 10
y = 5

print(x)
print(y)
```

```python
x = 5
print(x)

x = 10
print(x)

x = x + 3
print(x)
```

Kilka podstawowych elementów konwencji tworzenia zmiennych w Pythonie (nie wszystkie z nich są regułami):

- Nazwy zmiennych i funkcji pisz małymi literami, oddzielając wyrazy znakami podkreślenia (np. moja_zmienna, liczba_punktow),
- Wielkość liter: Python rozróżnia wielkość liter (Zmienna i zmienna to dwie różne nazwy),
- Cyfry: Nazwa może zawierać cyfry, ale nie może się od nich zaczynać (np. zmienna1 jest poprawna, 1zmienna jest błędna),
- Znaki specjalne: Używaj tylko liter, cyfr i podkreśleń _. Unikaj polskich znaków (np. ą, ś, ł) oraz spacji, 
- Słowa kluczowe: Nie używaj słów zastrzeżonych przez Pythona (np. if, for, class, def) - będziemy się o nich uczyć na tym kursie
- Stałe: Wartości stałe (które nie powinny się zmieniać) zapisuj w całości wielkimi literami, używając podkreśleń (np. MAKSYMALNA_PREDKOSC).

Dlaczego zapis `x = x + 3` jest poprawny w programowaniu, mimo że nie jest matematyczną równością?

## f-string
f-stringi (formatted string literals) pozwalają w prosty sposób umieszczać wartości zmiennych oraz wyniki obliczeń bezpośrednio w tekście.

```python
uczestnicy = 45
czas = 30

print(f"W badaniu uczestniczyło {uczestnicy} osób.")
print(f"Każdy uczestnik wykonywał zadania przez {czas} minut.")
print(f"Łączny czas udziału wszystkich osób: {uczestnicy * czas} minut.")
```

### Ćwiczenie 2. Kalkulator dwóch liczb

Napisz program, który:

1. Pobiera od użytkownika dwie liczby rzeczywiste `a` i `b` (zakładamy, że `b ≠ 0`).
2. Oblicza ich sumę, różnicę, iloczyn oraz iloraz.
3. Wyświetla wszystkie cztery wyniki wraz z opisami.


```python
a = float(input("Podaj pierwszą liczbę (a): "))
b = float(input("Podaj drugą liczbę (b): "))

suma = a + b
roznica = a - b
iloczyn = a * b
iloraz = a / b

print(f"Suma: ", suma)
print(f"Różnica: ", roznica)
print(f"Iloczyn: ", iloczyn)
print(f"Iloraz: ", iloraz)
```

### Ćwiczenie 3. Zamiana jednostek czasu

Napisz program, który:

1. Pobiera od użytkownika liczbę sekund `s` (liczba całkowita nieujemna).
2. Oblicza, ile pełnych godzin, minut i pozostałych sekund odpowiada podanej wartości.
3. Wyświetla wynik w postaci godzin, minut i sekund.


```python
s = int(input("Podaj liczbę sekund: "))

godziny = s // 3600
minuty = (s % 3600) // 60
sekundy = s % 60

print(f"Wprowadzona liczba sekund: {s}\n" )
print(f"Godziny:", godziny)
print(f"Minuty:", minuty)
print(f"Sekundy:", sekundy)
```

### Ćwiczenie 5. Koszt podróży

Napisz program, który:

1. Pobiera długość trasy w kilometrach `d`.
2. Pobiera średnie zużycie paliwa na 100 km `s`.
3. Pobiera cenę jednego litra paliwa `c`.
4. Oblicza ilość paliwa potrzebnego na przejazd oraz całkowity koszt podróży.
5. Wyświetla oba wyniki.

Zakładamy, że wszystkie dane wejściowe są liczbami rzeczywistymi dodatnimi.

```python
d = float(input("Podaj długość trasy (km): "))
s = float(input("Podaj średnie zużycie paliwa (l/100 km): "))
c = float(input("Podaj cenę paliwa (zł/l): "))

paliwo = (d / 100) * s
koszt = paliwo * c

print(f"Potrzebna ilość paliwa: {paliwo} l")
print(f"Całkowity koszt podróży: {koszt} zł")
```

### Ćwiczenie 6. Obliczanie BMI

Napisz program, który:

1. Pobiera masę ciała w kilogramach `m`.
2. Pobiera wzrost w centymetrach `h`.
3. Przelicza wzrost na metry.
4. Oblicza wskaźnik BMI według wzoru:

$$
BMI = \frac{m}{h_m^2}
$$

gdzie $h_m$ oznacza wzrost w metrach.

5. Wyświetla obliczoną wartość BMI.

```python
m = float(input("Podaj masę ciała (kg): "))
h = float(input("Podaj wzrost (cm): "))

h_m = h / 100
bmi = m / (h_m ** 2)

print("BMI wynosi:", bmi)
```

### Ćwiczenie 7. Analiza czasu reakcji

Napisz program, który:

1. Pobiera trzy czasy reakcji uczestnika eksperymentu w milisekundach: `t1`, `t2`, `t3`.
2. Oblicza średni czas reakcji.
3. Przelicza średni czas reakcji na sekundy.
4. Wyświetla oba wyniki.

Zakładamy, że wszystkie czasy są dodatnimi liczbami rzeczywistymi.

### Ćwiczenie 8. Wynik eksperymentu pamięciowego

W eksperymencie uczestnik zapamiętuje listę 20 słów.

Napisz program, który:

1. Pobiera liczbę poprawnie zapamiętanych słów `p` (liczba całkowita od 0 do 20).
2. Oblicza liczbę niezapamiętanych słów.
3. Oblicza procent poprawnie zapamiętanych słów.

### Ćwiczenie 9. Podział rachunku

Napisz program, który:

1. Pobiera całkowitą kwotę rachunku `k`.
2. Pobiera liczbę osób `n`.
3. Pobiera procent napiwku `p`.
4. Oblicza całkowitą kwotę wraz z napiwkiem.
5. Oblicza kwotę przypadającą na jedną osobę.
6. Wyświetla oba wyniki.

Zakładamy, że `k > 0`, `n` jest dodatnią liczbą całkowitą, a `p ≥ 0`.

```python
k = float(input("Podaj kwotę (zł): "))
n = int(input("Podaj liczbę osób: "))
p = float(input("Podaj procent napiwku: "))

napiwek = k * p / 100
kwota_calkowita = k + napiwek
na_osobe = kwota_calkowita / n

print(f"Całkowita kwota z napiwkiem: {kwota_calkowita} zł")
print(f"Kwota na jedną osobę: {na_osobe} zł")
```

### Ćwiczenie 10. Przetwarzanie danych eksperymentalnych

W eksperymencie uczestnik wykonał dwa bloki prób. W każdym bloku mierzono średni czas reakcji oraz liczbę wykonanych prób.

Napisz program, który:

1. Pobiera średni czas reakcji w pierwszym bloku `t1` oraz liczbę prób `n1`.
2. Pobiera średni czas reakcji w drugim bloku `t2` oraz liczbę prób `n2`.
3. Oblicza średni czas reakcji ze wszystkich prób, uwzględniając liczbę prób w każdym bloku.
4. Oblicza różnicę między średnimi czasami reakcji obu bloków (`t2 - t1`).
5. Wyświetla oba wyniki.

Zakładamy, że czasy są dodatnimi liczbami rzeczywistymi, a liczby prób są dodatnimi liczbami całkowitymi.

```python
t1 = float(input("Podaj czas reakcji w pierwszym bloku (ms): "))
n1 = int(input("Podaj liczbę prób w pierwszym bloku: "))

t2 = float(input("Podaj średni czas reakcji w drugim bloku (ms): "))
n2 = int(input("Podaj liczbę prób w drugim bloku: "))

srednia = (t1 * n1 + t2 * n2) / (n1 + n2)
roznica = t2 - t1

print(f"\nŚredni czas reakcji ze wszystkich prób: {srednia} ms")
print(f"Różnica między średnimi czasami reakcji: {roznica} ms")
```
