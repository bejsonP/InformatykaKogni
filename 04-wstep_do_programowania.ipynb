# Zmienne

**Zmienna** to nazwa, pod którą przechowywana jest wartość.

Operator `=` oznacza **przypisanie wartości**.

```python
x = 10
y = 5

print(x)
print(y)
```

```python
x = 5
print(x)

x = 10
print(x)

x = x + 3
print(x)
```

Kilka podstawowych elementów konwencji tworzenia zmiennych w Pythonie (nie wszystkie z nich są regułami):

- Nazwy zmiennych i funkcji pisz małymi literami, oddzielając wyrazy znakami podkreślenia (np. moja_zmienna, liczba_punktow),
- Wielkość liter: Python rozróżnia wielkość liter (Zmienna i zmienna to dwie różne nazwy),
- Cyfry: Nazwa może zawierać cyfry, ale nie może się od nich zaczynać (np. zmienna1 jest poprawna, 1zmienna jest błędna),
- Znaki specjalne: Używaj tylko liter, cyfr i podkreśleń _. Unikaj polskich znaków (np. ą, ś, ł) oraz spacji, 
- Słowa kluczowe: Nie używaj słów zastrzeżonych przez Pythona (np. if, for, class, def) - będziemy się o nich uczyć na tym kursie
- Stałe: Wartości stałe (które nie powinny się zmieniać) zapisuj w całości wielkimi literami, używając podkreśleń (np. MAKSYMALNA_PREDKOSC).
