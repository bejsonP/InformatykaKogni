# Instrukcja if

Instrukcja `if` wykonuje wskazany fragment kodu **tylko wtedy**, gdy warunek jest prawdziwy.

Po warunku umieszczamy dwukropek `:`, a polecenia wewnątrz bloku zapisujemy z wcięciem (zwykle cztery spacje).

```python
a = 15
b = 10

if a > b:
    print(a)
        
print("To wykona się niezależnie od pętli")
```

```python
wiek = 20

if wiek >= 18:
    print("Osoba pełnoletnia")

print("To wykona się niezależnie od pętli")
```

# Instrukcja if...else

`else` określa, co wykonać, jeśli warunek w `if` jest fałszywy.

```python
wynik = 100

if wynik >= 50:
    print("Zaliczone")
else:
    print("Niezaliczone")


print("To wykona się niezależnie od pętli")
```

# Instrukcja if...elif...else

`elif` pozwala sprawdzać kolejne warunki. Wykonany zostaje **pierwszy pasujący blok**.

```python
punkty = 80

if punkty >= 90: # False
    print("Bardzo dobry wynik")
    print("Gratulacje")
elif punkty >= 60:  # True
    print("Dobry wynik")
elif punkty >= 40: # True
    print("Wynik przeciętny")
else:
    print("Wynik nieprzewidziany")

print("To wykona się niezależnie od pętli")
```

# Operatory logiczne

| Operator | Znaczenie | Przykład |
|---|---|---|
| `and` | oba warunki prawdziwe | `wiek >= 18 and wiek <= 65` |
| `or` | przynajmniej jeden warunek prawdziwy | `x < 0 or x > 100` |
| `not` | zaprzeczenie | `not czy_student` |

```python
temperatura = 23
czy_pada = False

if temperatura >= 20 and not czy_pada:
    print("Można iść na spacer")
else:
    print("Lepiej zostać w domu")
```

```python
czy_weekend = False
czy_urlop = True

if czy_weekend or czy_urlop:
    print("Dzisiaj nie trzeba iść do pracy")
else:
    print("Dzisiaj trzeba iść do pracy")
```

```python
czy_sklep_otwarty = False

if not czy_sklep_otwarty:
    print("Sklep jest zamknięty")
else:
    print("Można zrobić zakupy")
```

```python
godzina = 17
czy_weekend = True
czy_pada = False

if (godzina <= 21 or czy_weekend) and not czy_pada:
    print("Można wyjść na spacer")
else:
    print("Lepiej zostać w domu")
```

# Zagnieżdżone instrukcje warunkowe

Instrukcja `if` może znajdować się wewnątrz innego `if`. Czasem jednak czytelniej połączyć warunki za pomocą `and`.

```python
wiek = 21
zgoda = True

if wiek >= 18:
    if zgoda:
        print("Można rozpocząć eksperyment")
    else:
        print("Brak zgody")
else:
    print("Uczestnik niepełnoletni")
```

Jak można zapisać powyższy kod z użyciem `if` z użyciem `and`?

```python
wiek = 21
zgoda = True

if wiek >= 18 and zgoda:
    print("Można rozpocząć eksperyment")
elif wiek < 18:
    print("Uczestnik niepełnoletni")
else:
    print("Brak zgody")
```

# Typowe błędy

- Używanie `=` zamiast `==` w porównaniach.
- Pominięcie dwukropka po `if`, `elif` lub `else`.
- Nieprawidłowe wcięcia.
- Porównywanie tekstu z liczbą bez konwersji.
- Zła kolejność warunków `elif`.

### Ćwiczenie 5. Prosty kalkulator

Napisz program, który:

Pobiera dwie liczby oraz znak działania (`+`, `-`, `*`, `/`). Wykonuje wybrane działanie. W przypadku dzielenia przez zero wyświetla odpowiedni komunikat.

```python
a = float(input("Podaj pierwszą liczbę: "))
b = float(input("Podaj drugą liczbę: "))
dzialanie = input("Podaj działanie (+, -, *, /): ")

if dzialanie == "+":
    print("Wynik: ", a + b)

elif dzialanie == "-":
    print("Wynik: ", a - b)

elif dzialanie == "*":
    print("Wynik: :", a * b)

elif dzialanie == "/":

    if b != 0:
        print("Wynik: ", a / b)
    else:
        print("Błąd: nie można dzielić przez zero")
else:
    print("Nieznane działanie")
```

### Zadanie 10. Trójkąt

Pobierz trzy długości boków. Najpierw sprawdź, czy są dodatnie i czy można z nich utworzyć trójkąt. Jeśli tak, ustal, czy jest równoboczny, równoramienny czy różnoboczny. Przetestuj 3,4,5; 2,2,3; 1,2,3.

```python
a = float(input("Podaj długość boku a: "))
b = float(input("Podaj długość boku b: "))
c = float(input("Podaj długość boku c: "))

if a <= 0 or b <= 0 or c <= 0:
    print("Długości boków muszą być dodatnie")

elif a + b <= c or a + c <= b or b + c <= a:
    print("Z podanych boków nie można utworzyć trójkąta")

else:
    print("Z podanych boków można utworzyć trójkąt")
    if a == b == c:
        print("Trójkąt równoboczny")
    elif a == b or a == c or b == c:
        print("Trójkąt równoramienny")
    else:
        print("Trójkąt różnoboczny")
```

### Zadanie 11. Gra: zgadnij wynik

Ustal w kodzie liczbę `sekret = 37`. Użytkownik ma **dokładnie trzy próby**, wczytywane osobnymi instrukcjami (bez pętli). Po każdej próbie wyświetl `za mało`, `za dużo` albo `trafione`. Jeśli odgadnie wcześniej, nie pytaj o kolejne próby.

```python
sekret = 37

proba1 = int(input("Próba 1. Podaj liczbę: "))

# Pierwsza próba
if proba1 == sekret:
    print("Trafione!")

else:
    if proba1 < sekret:
        print("Za mało")
    else:
        print("Za dużo")

    # Druga próba
    proba2 = int(input("Próba 2. Podaj liczbę: "))

    if proba2 == sekret:
        print("Trafione!")
    else:
        if proba2 < sekret:
            print("Za mało")
        else:
            print("Za dużo")

        proba3 = int(input("Próba 3. Podaj liczbę: "))

        # Trzecia próba
        if proba3 == sekret:
            print("Trafione!")
        elif proba3 < sekret:
            print("Za mało")
        else:
            print("Za dużo")

        if proba3 != sekret:
            print(f"Koniec gry! Szukana liczba to {sekret}.")
```
